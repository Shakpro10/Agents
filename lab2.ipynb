```python
# Import relevant libraries

import os
import json
from dotenv import load_dotenv
from openai import OpenAI

from IPython.display import Markdown, display
```

```python
# Load the API keys into environment variables
load_dotenv(override=True)
```

```python
# Print the key prefixes to help with any debugging

openai_api_key = os.getenv('NVIDIA_API_KEY_1')
google_api_key = os.getenv('GEMINI_API_KEY')
nemotron_api_key = os.getenv('NVIDIA_API_KEY_2')
groq_api_key = os.getenv('GROQ_API_KEY')

if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:4]}")
else:
    print("OpenAI API Key not set")

if google_api_key:
    print(f"Google API Key exists and begins {google_api_key[:4]}")
else:
    print("Google API Key not set (and this is optional)")

if nemotron_api_key:
    print(f"Nemotron API Key exists and begins {nemotron_api_key[:4]}")
else:
    print("Nemotron API Key not set (and this is optional)")

if groq_api_key:
    print(f"Groq API Key exists and begins {groq_api_key[:4]}")
else:
    print("Groq API Key not set (and this is optional)")
```

```python
request = "Please come up with a challenging, nuanced question that I can ask a number of LLMs to evaluate their intelligence. "
request += "Answer only with the question, no explanation."
messages = [{"role": "user", "content": request}]

messages
```

```python
openai = OpenAI(
    base_url = "https://integrate.api.nvidia.com/v1", 
    api_key = openai_api_key
)
response_1 = openai.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=messages,
)

question_1 = response_1.choices[0].message.content
print(question_1)
```

```python
display(Markdown(question_1))
```

```python
# Store the competitors and their answers in lists for later use to compare 
# the answers and determine a winner
competitors = []
answers = []
messages = [{"role": "user", "content": question_1}]
```

```python
# Now we will ask the same question to a number of different LLMs and see how 
# they do. We will start with the same model, but this time we will ask it to 
# answer the question instead of coming up with the question.

model_name = "openai/gpt-oss-20b"

response_2 = openai.chat.completions.create(model=model_name, messages=messages)
answer_1 = response_2.choices[0].message.content

display(Markdown(answer_1))
competitors.append(model_name)
answers.append(answer_1)
```

```python
# I've updated this with the latest model (gemini-2.5-flash), but it can take some time because it 
# likes to think!

gemini = OpenAI(api_key=google_api_key, base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
model_name_2 = "gemini-2.5-flash"

response_3 = gemini.chat.completions.create(model=model_name_2, messages=messages)
answer_2 = response_3.choices[0].message.content

display(Markdown(answer_2))
competitors.append(model_name_2) 
answers.append(answer_2)
```

```python
# I've updated this with the nemotron-3-super-120b-a12b model

nemotron = OpenAI(api_key=nemotron_api_key, base_url="https://integrate.api.nvidia.com/v1")
model_name_3 = "nvidia/nemotron-3-super-120b-a12b"

response_4 = nemotron.chat.completions.create(
    model=model_name_3, 
    messages=messages)
answer_3 = response_4.choices[0].message.content

display(Markdown(answer_3))
competitors.append(model_name_3)
answers.append(answer_3)
```

```python
# I've updated this with the llama-3.3-70b-versatile model

groq = OpenAI(api_key=groq_api_key, base_url="https://api.groq.com/openai/v1")
model_name_4 = "llama-3.3-70b-versatile"

response_5 = groq.chat.completions.create(model=model_name_4, messages=messages)
answer_4 = response_5.choices[0].message.content

display(Markdown(answer_4))
competitors.append(model_name_4)
answers.append(answer_4)
```

```python
# Updated with the open source model llama3.2 using ollama

ollama = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
model_name_5 = "llama3.2"

response_6 = ollama.chat.completions.create(model=model_name_5, messages=messages)
answer_5 = response_6.choices[0].message.content

display(Markdown(answer_5))
competitors.append(model_name_5)
answers.append(answer_5)
```

```python
print(competitors)
print(answers)
```

```python
# Zip the competitors and answers together to compare them side by side. 
# This is a simple way to do it, but you could also use a more sophisticated 
# method like a table or a dataframe.

for competitor, answer in zip(competitors, answers):
    print(f"Competitor: {competitor}\n\n{answer}")
```

```python
# Bring both answers and competitors together

together = ""
for index, answer in enumerate(answers):
    together += f"# Response from competitor {index+1}\n\n"
    together += answer + "\n\n"
```

```python
print(together)
```

#### This next section test the intelligence of each LLM queried above by comparing their answers. This test is carried out by a separate LLM

```python
# Now we will judge the competitors based on their answers to the question. 
# We will use the a stronger gpt model to judge the answers. 

judge = f"""You are judging a competition between {len(competitors)} competitors.
Each model has been given this question:

{question_1}

Your job is to evaluate each response for clarity and strength of argument, and rank them in order of best to worst.
Respond with JSON, and only JSON, with the following format:
{{"results": ["best competitor number", "second best competitor number", "third best competitor number", ...]}}

Here are the responses from each competitor:

{together}

Now respond with the JSON with the ranked order of the competitors, nothing else. Do not include markdown formatting or code blocks."""
```

```python
print(judge)
```

```python
judge_messages = [{"role": "user", "content": judge}]
```

```python
# Judgement time!

gpt = OpenAI(api_key=openai_api_key, base_url="https://integrate.api.nvidia.com/v1")
model_name_6 = "openai/gpt-oss-120b"

response_7 = gpt.chat.completions.create(model=model_name_6, messages=judge_messages)
results = response_7.choices[0].message.content
print(results)
```

```python
# OK let's turn this into results!

results_dict = json.loads(results)
ranks = results_dict["results"]
for index, result in enumerate(ranks):
    competitor = competitors[int(result)-1]
    print(f"Rank {index+1}: {competitor}")
```
