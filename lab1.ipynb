```python
from dotenv import load_dotenv

# Load the API keys into environment variables
load_dotenv(override=True)
```

```python
# Check the key - if you're not using OpenAI, check whichever key you're using! Ollama doesn't need a key.

import os

GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
google_api_key = os.getenv('GOOGLE_API_KEY')

if google_api_key:
    print(f"Gemini API Key exists and begins {google_api_key[:8]}")
else:
    print("Gemini API Key not set. Please set the GOOGLE_API_KEY environment variable.")
    
```

```python
from openai import OpenAI
from IPython.display import Markdown, display

# Create an instance of the gemini class using openAI endpoint and the google api key
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)
```

```python
# First create the messages:
question = 'Pick a business area that might be worth exploring for an Agentic AI opportunity. Respond briefly and concisely'

messages = [{"role": "user", "content": question}]

# Make the first call:
  
response = gemini.chat.completions.create(
    model="gemini-2.5-flash-lite",
    messages=messages
)

# Then read the business idea:

business_idea = response.choices[0].message.content
display(Markdown(business_idea))
```

```python
# Second task
question_2 = f"Present a pain-point in {business_idea} for Agentic AI opportunity - something challenging that might be ripe for an Agentic solution. Respond briefly and concisely."

messages = [{"role": "user", "content": question_2}]

# Then make the first call:
  
response = gemini.chat.completions.create(
    model="gemini-2.5-flash-lite",
    messages=messages
)

# Then read the business idea:

pain_point = response.choices[0].message.content

display(Markdown(pain_point))
```

```python
# 3rd task
question_3 = f"""
Given the following pain point in an industry:

{pain_point}

Propose a clear Agentic AI solution that:
- uses autonomous multi-step reasoning
- explains how agents would solve the problem
- is practical and implementable

Respond briefly and clearly.
"""

messages = [{"role": "user", "content": question_3}]

# Then make the first call:
  
response = gemini.chat.completions.create(
    model="gemini-2.5-flash",
    messages=messages
)

# Then read the business idea:

answer = response.choices[0].message.content
display(Markdown(answer))

# And repeat! In the next message, include the business idea within the message
```
